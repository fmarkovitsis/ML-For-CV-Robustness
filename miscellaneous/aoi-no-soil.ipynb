```python
import numpy as np
import OpenEXR
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image, ImageDraw
```

```python
def read_exr_rgb(path):
    with OpenEXR.File(path) as exr:
        part = exr.parts[0]
        channels = part.channels
        rgb = channels["RGBA"].pixels[..., :3].astype(np.float32)
    return rgb
```

```python
def get_blocks(img):
    blocks = []
    
    for i in range (0, 20):
        block = img[starting_y:starting_y+x_step, starting_x+(i*x_step):starting_x+(i+1)*x_step]
        blocks.append(block)

    return blocks
```

```python
imageFolder = Path("/home/philipm/Desktop/imagesTest")
imageFolder.is_dir()
```

```python
path = "/home/philipm/Desktop/proc/"
```

```python
counter = 0

for file in imageFolder.glob("*.exr"):
    file = str(file)

    img = read_exr_rgb(file)

    y, x, c = img.shape
    
    x_step = x//24
    starting_x = 2 * x_step
    
    starting_y = y//2-x_step//2

    blocks = get_blocks(img)
    blocks_np = np.stack(blocks)

    np.save(path + f"proc_{counter}.npy", blocks_np)

    counter = counter + 1
```
