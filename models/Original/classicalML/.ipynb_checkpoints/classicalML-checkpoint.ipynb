```python
import cuml

from cuml.ensemble import RandomForestRegressor

from sklearn import ensemble
from sklearn.ensemble import ExtraTreesRegressor

import sklearn

import matplotlib.pyplot as plt

import xgboost as xgb
from lightgbm import LGBMRegressor

import torch

import numpy as np
import cupy as cp

import pandas as pd
from pathlib import Path
import gc

from sklearn.metrics import r2_score

import time
```

```python
randomf = True
extrat = True
xgboost = True
lgboostm = True
```

```python
modelsLib = {
    "rf": 1,
    "rfGPU": 1,
    "et": 1,
    "etGPU": 1,
    "xgb": 1,
    "xgbGPU": 1,
    "lgbm": 1,
    "lgbmGPU": 1
}
```

```python
# RF

RF = {
    "model": RandomForestRegressor,
    "n_estimators": [50, 50, 50],
    "max_depth": [24, 24, 1],
    "n_bins": [256, 256, 2],
    "max_features": [1.0, 1.0, 0.1],
    "results": [],
    "trials": []
}
```

```python
# ET

ET = {
    "model": ExtraTreesRegressor,
    "n_estimators": [50, 50, 50],
    "max_depth": [21, 21, 1],
    "min_samples_leaf": [4, 4, 2],
    "max_features": [1.0, 1.0, 0.1],
    "results": []
}
```

```python
datasetPath = Path("/home/philipm/Desktop/ML-For-CV-Robustness/datasets/notrees-dl.csv")

#making sure
datasetPath.is_file()
```

```python
dataset = pd.read_csv(datasetPath)
dataset.head()
```

```python
target_col = dataset["grviOut"]
labels = target_col.to_numpy(dtype="float32")
labels
```

```python
inputs = dataset.drop("grviOut", axis=1)
inputs = inputs.drop("sunRoll", axis=1)
inputs = inputs.drop("camRoll", axis=1)

inputs.head()
```

```python
inputs = inputs.to_numpy(dtype="float32")
inputs
```

```python
if len(labels) != len (inputs):
    print ("problem")
```

```python
n_total = len(inputs)

n_train = int(0.9*n_total)
n_val = int(0.1*n_total)

# shuffle
g = torch.Generator().manual_seed(42)
perm = torch.randperm(n_total, generator=g).numpy()

inputs_shuf = inputs[perm]
labels_shuf = labels[perm]

# split
train_inputs, train_labels = inputs_shuf[:n_train], labels_shuf[:n_train]
val_inputs, val_labels = inputs_shuf[n_train:n_train + n_val], labels_shuf[n_train:n_train + n_val]
```

```python
# scaling
from sklearn.preprocessing import QuantileTransformer, StandardScaler

if False:
    input_scaler = StandardScaler() #QuantileTransformer(output_distribution='normal')
    target_scaler = StandardScaler()

    train_inputs = input_scaler.fit_transform(train_inputs)
    train_labels = target_scaler.fit_transform(train_labels.reshape(-1, 1)).reshape(-1)

    val_inputs = input_scaler.transform(val_inputs)
    val_labels = target_scaler.transform(val_labels.reshape(-1, 1)).reshape(-1)
```

```python
# RF
# starting point
if randomf:
    
    ne = RF["n_estimators"][0]
    md = RF["max_depth"][0]
    nb = RF["n_bins"][0]
    mf = RF["max_features"][0]
    
    while True:
        if ne > RF["n_estimators"][1]:
            ne = RF["n_estimators"][0]
            break
    
        while True:
            if md > RF["max_depth"][1]:
                md = RF["max_depth"][0]
                break
    
            while True:
                if nb > RF["n_bins"][1]:
                    nb = RF["n_bins"][0]
                    break
    
                while True:
                    if mf > RF["max_features"][1]:
                        mf = RF["max_features"][0]
                        break
    
                    modelrfGPU = RF["model"](n_estimators=ne, max_depth=md, n_bins=nb, max_features=mf)
                    modelrfGPU.fit(train_inputs, train_labels)
    
                    y_pred = modelrfGPU.predict(val_inputs)
                    modelsLib["rfGPU"] = modelrfGPU
                    
                    mse = cuml.metrics.mean_squared_error(val_labels, y_pred)
                    rmse = cp.sqrt(mse)
    
                    RF["results"].append(float(rmse))
                    RF["trials"].append({
                        "n_estimators": ne,
                        "max_depth": md,
                        "n_bins": nb,
                        "max_features": mf,
                        "rmse": float(rmse)
                    })
    
                    mf = mf + RF["max_features"][2]
    
                nb = nb * RF["n_bins"][2]
    
            md = md + RF["max_depth"][2]
    
        ne = ne+ RF["n_estimators"][2]

    
    r2 = r2_score(val_labels, y_pred)
    rel_test_loss = min(RF["results"])*100 / (val_labels.max()-val_labels.min())
    print(str(float(rel_test_loss)) + "%")
    print ("r2: " + str(r2))
    print ("rmse: " + str(rmse))
    


    plt.plot(range(len(RF["results"])), RF["results"])
    plt.xlabel("Index")
    plt.ylabel("Value")
    plt.show()
```

```python
modelrf = ensemble.RandomForestRegressor(n_estimators=ne, max_depth=md, max_features=mf)
modelrf.fit(train_inputs, train_labels)

y_pred = modelrf.predict(val_inputs)
modelsLib["rf"] = modelrf

mse = sklearn.metrics.mean_squared_error(val_labels, y_pred)
rmse = cp.sqrt(mse)


r2 = r2_score(val_labels, y_pred)
rel_test_loss = (rmse*100) / (val_labels.max()-val_labels.min())
print(str(float(rel_test_loss)) + "%")
print ("r2: " + str(r2))
print ("rmse: " + str(rmse))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

if randomf:

    df = pd.DataFrame(RF["trials"])
    
    params = [
        "n_estimators",
        "max_depth",
        "n_bins",
        "max_features"
    ]
    
    # Find best configuration
    best = df.loc[df["rmse"].idxmin()]
    
    fig, axes = plt.subplots(2, 2, figsize=(12, 8))
    
    for ax, param in zip(axes.flatten(), params):
    
        # Keep the other three hyperparameters fixed
        mask = np.ones(len(df), dtype=bool)
    
        for other in params:
            if other != param:
                mask &= df[other] == best[other]
    
        subset = df[mask].sort_values(param)
    
        ax.plot(
            subset[param],
            subset["rmse"],
            marker="o",
            linewidth=1.5
        )
    
        ax.set_xlabel(param)
        ax.set_ylabel("Validation RMSE")
        ax.grid(alpha=0.3)
        
    ymin = min(ax.get_ylim()[0] for ax in axes.flatten())
    ymax = max(ax.get_ylim()[1] for ax in axes.flatten())
    
    for ax in axes.flatten():
        ax.set_ylim(ymin, ymax)
        
    plt.suptitle(
        "Random Forest Hyperparameter Sensitivity",
        fontsize=14
    )
    plt.savefig("/home/philipm/Desktop/gridSeachRF2nd-sepcs", dpi=300, bbox_inches="tight")
    plt.tight_layout()
    plt.show()
```

```python
# XGB
# starting point

if xgboost:

    ne = X["n_estimators"][0]
    md = X["max_depth"][0]
    mc = X["min_child_weight"][0]
    
    while True:
        if ne > X["n_estimators"][1]:
            ne = X["n_estimators"][0]
            break
    
        while True:
            if md > X["max_depth"][1]:
                md = X["max_depth"][0]
                break
    
            while True:
                if mc > X["min_child_weight"][1]:
                    mc = X["min_child_weight"][0]
                    break
    
                modelXGB = X["model"].XGBRegressor(n_estimators=ne, max_depth=md, min_child_weight=mc, learning_rate=0.05, tree_method="gpu_hist", gpu_id=0)
                modelXGB.fit(train_inputs, train_labels)

                y_pred = modelXGB.predict(val_inputs)

                modelsLib["xgb"] = modelXGB

                mse = sklearn.metrics.mean_squared_error(val_labels, y_pred)
                rmse = np.sqrt(mse)

                X["results"].append(float(rmse))
    
                mc = mc + X["min_child_weight"][2]
    
            md = md + X["max_depth"][2]
    
        ne = ne+ X["n_estimators"][2]
    
        rel_test_loss = min(X["results"])*100 / (val_labels.max()-val_labels.min())
        print(str(float(rel_test_loss)) + "%")


        r2 = r2_score(val_labels, y_pred)
        print ("r2: " + str(r2))
        print ("rmse: " + str(rmse))
        
        plt.plot(range(len(X["results"])), X["results"])
        plt.xlabel("Index")
        plt.ylabel("Value")
        plt.show()
```

```python
LGBM = {
    "model": LGBMRegressor,
    "n_estimators": [450, 450, 50],
    "min_child_samples": [15, 15, 10],
    "num_leaves": [256, 256, 2],
    "colsample_bytree": [0.7, 0.7, 0.1],
    "results": [],
    "trials": []
}
```

```python
if lgboostm:

    ne = LGBM["n_estimators"][0]
    mcs = LGBM["min_child_samples"][0]
    nl = LGBM["num_leaves"][0]
    cb = LGBM["colsample_bytree"][0]
    
    while True:
        if ne > LGBM["n_estimators"][1]:
            ne = LGBM["n_estimators"][0]
            break
    
        while True:
            if mcs > LGBM["min_child_samples"][1]:
                mcs = LGBM["min_child_samples"][0]
                break
    
            while True:
                if nl > LGBM["num_leaves"][1]:
                    nl = LGBM["num_leaves"][0]
                    break
    
                while True:
                    if cb > LGBM["colsample_bytree"][1]:
                        cb = LGBM["colsample_bytree"][0]
                        break
    
                    modelLGBM = LGBM["model"](n_estimators=ne, min_child_samples=mcs, num_leaves=nl-1, colsample_bytree=cb)
                    modelLGBM.fit(train_inputs, train_labels)
    
                    y_pred = modelLGBM.predict(val_inputs)

                    modelsLib["lgbm"] = modelLGBM
                    
                    mse = sklearn.metrics.mean_squared_error(val_labels, y_pred)
                    rmse = np.sqrt(mse)
    
                    LGBM["results"].append(float(rmse))

                    LGBM["trials"].append({
                        "n_estimators": ne,
                        "min_child_samples": mcs,
                        "num_leaves": nl,
                        "colsample_bytree": cb,
                        "rmse": float(rmse)
                    })
    
                    cb = round(cb + LGBM["colsample_bytree"][2], 1)
    
                nl = nl * LGBM["num_leaves"][2]
    
            mcs = mcs + LGBM["min_child_samples"][2]
    
        ne = ne + LGBM["n_estimators"][2]
    
    rel_test_loss = min(LGBM["results"])*100 / (val_labels.max()-val_labels.min())
    print(str(float(rel_test_loss)) + "%")


    r2 = r2_score(val_labels, y_pred)
    print ("r2: " + str(r2))
    print ("rmse: " + str(rmse))
```

```python
import time
import cupy as cp
import numpy as np
import cuml

modelGPU = modelsLib["rfGPU"]

x_cpu = np.asarray(val_inputs[0:1], dtype=np.float32)

def measure(fn):
    cp.cuda.Device().synchronize()
    start = time.perf_counter()

    result = fn()

    cp.cuda.Device().synchronize()
    return result, (time.perf_counter() - start) * 1000

with cuml.using_output_type("cupy"):

    for i in range(0, 20):
        x_gpu = cp.asarray(val_inputs[i:i+1], dtype=cp.float32)
        pred = modelGPU.predict(x_gpu)
        pred_cpu = cp.asnumpy(pred)


with cuml.using_output_type("cupy"):

    x_gpu, t_transfer = measure(
        lambda: cp.asarray(x_cpu)
    )

    pred, t_predict = measure(
        lambda: modelGPU.predict(x_gpu)
    )

    pred_cpu, t_output = measure(
        lambda: cp.asnumpy(pred)
    )

print(f"CPU → GPU: {t_transfer:.4f} ms")
print(f"Prediction: {t_predict:.4f} ms")
print(f"GPU → CPU: {t_output:.4f} ms")
```

```python
import time
import numpy as np
import cupy as cp
import cuml

modelGPU = modelsLib["rfGPU"]

X = np.asarray(val_inputs, dtype=np.float32)

with cuml.using_output_type("cupy"):

    for batch_size in [1, 32, 512, 4096]:

        if len(X) < batch_size:
            continue

        x_gpu = cp.asarray(X[:batch_size])

        # Warm-up
        for _ in range(5):
            _ = modelGPU.predict(x_gpu)

        cp.cuda.Device().synchronize()

        host_times = []
        sync_times = []

        for _ in range(5):

            cp.cuda.Device().synchronize()

            t0 = time.perf_counter()

            pred = modelGPU.predict(x_gpu)

            t1 = time.perf_counter()

            cp.cuda.Device().synchronize()

            t2 = time.perf_counter()

            host_times.append((t1 - t0) * 1000)
            sync_times.append((t2 - t1) * 1000)

        print(f"\nBatch size: {batch_size}")
        print(f"Predict call: {np.median(host_times):.4f} ms")
        print(f"Final sync:   {np.median(sync_times):.4f} ms")
        print(
            f"Total:        "
            f"{np.median(np.array(host_times) + np.array(sync_times)):.4f} ms"
        )
```

```python
import pandas as pd
import pickle

from sklearn.ensemble import (
    RandomForestRegressor,
    RandomForestClassifier,
    ExtraTreesRegressor,
    ExtraTreesClassifier
)

from xgboost import XGBModel
from lightgbm import LGBMModel


def get_model_complexity(model):

    # Random Forest / Extra Trees (sklearn)
    if isinstance(model, (
        RandomForestRegressor,
        RandomForestClassifier,
        ExtraTreesRegressor,
        ExtraTreesClassifier
    )):

        trees = len(model.estimators_)

        leaves = sum(
            tree.tree_.n_leaves
            for tree in model.estimators_
        )

        nodes = sum(
            tree.tree_.node_count
            for tree in model.estimators_
        )

    # XGBoost
    elif isinstance(model, XGBModel):

        df = model.get_booster().trees_to_dataframe()

        trees = df["Tree"].nunique()
        leaves = (df["Feature"] == "Leaf").sum()
        nodes = len(df)

    # LightGBM
    elif isinstance(model, LGBMModel):

        df = model.booster_.trees_to_dataframe()

        trees = df["tree_index"].nunique()
        leaves = df["split_feature"].isna().sum()
        nodes = len(df)

    else:
        raise TypeError(
            f"Unsupported model type: {type(model).__name__}"
        )

    # Serialized model size in MiB
    model_size = len(pickle.dumps(model)) / (1024 ** 2)

    return {
        "Trees": int(trees),
        "Leaves": int(leaves),
        "Nodes": int(nodes),
        "Leaves/Tree": round(leaves / trees, 2),
        "Nodes/Tree": round(nodes / trees, 2),
        "Size (MiB)": round(model_size, 3)
    }


# =====================================================
# EXTRACT COMPLEXITY METRICS FROM modelsLib
# =====================================================

results = {}

for name, model in modelsLib.items():

    try:
        results[name] = get_model_complexity(model)

    except (TypeError, AttributeError, ValueError) as e:
        print(f"Skipping {name}: {e}")


# Create comparison DataFrame
complexity_df = pd.DataFrame.from_dict(
    results,
    orient="index"
)

complexity_df.index.name = "Model"

display(complexity_df)


# Optional: export results
complexity_df.to_csv("model_complexity.csv")
```
