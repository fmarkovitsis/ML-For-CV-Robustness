```python
import numpy as np
import pandas as pd
import OpenEXR
import re
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image, ImageDraw
```

```python
csvPath = Path(r"C:\Users\sol77\Desktop\ML-For-CV-Robustness\miscellaneous\grviValues.csv")
csvPath.is_file()
```

```python
grvi_gt_raw = pd.read_csv(csvPath)
grvi_gt_raw
```

```python
cameraPathPath = Path(r"C:\Users\sol77\Desktop\ML-For-CV-Robustness\miscellaneous\ObjectPoses_back.txt")
```

```python
with open(str(cameraPathPath), "r") as f:
     cameraPathFile = f.readlines()
```

```python
cameraCoords = []
cameraRot = []

for line in cameraPathFile:
    xyz = []
    
    numbers = re.findall(r'-?\d+(?:\.\d+)?(?:[eE][+-]?\d+)?', line)

    xyz.append(float(numbers[1]))
    xyz.append(float(numbers[2]))

    cameraCoords.append(xyz)
    cameraRot.append(float(numbers[6]))
```

```python
jsks = []

for i in range (0, len(cameraCoords)):
    coords = cameraCoords[i]

    if cameraRot[i] == 0:
        j = 0
        while True:
            if coords[0]+400 < ((-14840) + j*1900):
                break
                
            j = j + 1
        j = j - 1
        
        for l in range (0, 20):
            k = 0
            while True:
                if coords[1] - 340 + (l+0.5)*(680/20) < ((-14780) + k*1900):
                    break
    
                k = k + 1
            k = k - 1
            if (k == 13):
                k = 12

            jsks.append([j, k])
    
    elif cameraRot[i] == -180:
        j = 0
        while True:
            if coords[0]-400 < ((-14840) + j*1900):
                break
                
            j = j + 1
        j = j - 1
        
        for l in range (0, 20):
            k = 0
            while True:
                if coords[1] + 340 - (l+0.5)*(680/20) < ((-14780) + k*1900):
                    break
    
                k = k + 1
            k = k - 1
            if (k == 13):
                k = 12

            jsks.append([j, k]) 
    else:
        j = 0
        while True:
            if coords[0] < ((-14840) + j*1900):
                break
                
            j = j + 1
        j = j - 1

        for l in range (0, 20):
            k = 0
            while True:
                if coords[1] + 400 < ((-14780) + k*1900):
                    break
    
                k = k + 1
            k = k - 1
            if (k == 13):
                k = 12

            jsks.append([j, k])
        
jsks
```

```python
len(jsks)
```

```python
grviOut = []

for i in range (0, len(jsks)):
    jk = jsks[i]
    if (jk[1] == 13):
        print (i)
        break
    
    grviOut.append(grvi_gt_raw.iloc[13-jk[0], jk[1]])
```

```python
len(grviOut)
```

```python
data = pd.DataFrame({"grviOut": grviOut})
```
