```python
import numpy as np
import OpenEXR
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image, ImageDraw
```

```python
# define camera and sun angles

c =  np.deg2rad(np.array([0.0, -19.0,  0.0]))
s =  np.deg2rad(np.array([10.0, 85.0, 90.0]))
```

```python
# calculate rotation matrices

# camera
c_r_cos = np.cos(c[0])
c_r_sin = np.sin(c[0])

c_p_cos = np.cos(c[1])
c_p_sin = np.sin(c[1])

c_y_cos = np.cos(c[2])
c_y_sin = np.sin(c[2])

# sun
s_r_cos = np.cos(s[0])
s_r_sin = np.sin(s[0])

s_p_cos = np.cos(s[1])
s_p_sin = np.sin(s[1])

s_y_cos = np.cos(s[2])
s_y_sin = np.sin(s[2])
```

```python
#camera
Rxc = np.array([[1.0, 0.0, 0.0], [0.0, c_r_cos, -1.0 * c_r_sin], [0.0, c_r_sin, c_r_cos]])
Ryc = np.array([[c_p_cos, 0.0, c_p_sin], [0.0, 1.0, 0.0], [-1*c_p_sin, 0.0, c_p_cos]])
Rzc = np.array([[c_y_cos, -1*c_y_sin, 0.0], [c_y_sin, c_y_cos, 0.0], [0.0, 0.0, 1]])

#sun
Rxs = np.array([[1.0, 0.0, 0.0], [0.0, s_r_cos, -1.0 * s_r_sin], [0.0, s_r_sin, s_r_cos]])
Rys = np.array([[s_p_cos, 0.0, s_p_sin], [0.0, 1.0, 0.0], [-1*s_p_sin, 0.0, s_p_cos]])
Rzs = np.array([[s_y_cos, -1*s_y_sin, 0.0], [s_y_sin, s_y_cos, 0.0], [0.0, 0.0, 1]])
```

```python
Rc = np.matmul(Rzc, np.matmul(Ryc, Rxc))
Rs = np.matmul(Rzs, np.matmul(Rys, Rxs))
```

```python
f = np.array([1.0, 0.0, 0.0])
r = np.array([0.0, 1.0, 0.0])
```

```python
fc = np.matmul(Rc, f)
rc = np.matmul(Rc, r)
fs = np.matmul(Rs, f)
```

```python
sf = np.dot(fs, fc)
sr = np.dot(fs, rc)
```

```python
angle = np.arctan2(sr, sf)
```
