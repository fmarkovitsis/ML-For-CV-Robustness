```python
import numpy as np
import pandas as pd
import OpenEXR
import re
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image, ImageDraw
```

```python
gt_front_path = Path(r"D:\ThesisData\front_notrees\1020\GroundTruth\data.csv")
gt_back_path = Path(r"D:\ThesisData\back_notrees\1020\GroundTruth\data.csv")

frontPath = Path(r"D:\ThesisData\front_notrees")
backPath = Path(r"D:\ThesisData\back_notrees")
```

```python
gt_front = pd.read_csv(gt_front_path)
gt_front = gt_front[["grviIn"]]
gt_front = gt_front.rename(columns={"grviIn": "grviOut"})

gt_back = pd.read_csv(gt_back_path)
gt_back = gt_back[["grviIn"]]
gt_back = gt_back.rename(columns={"grviIn": "grviOut"})

dataset = pd.DataFrame()
```

```python
for folder in frontPath.glob("*"):
    for subfolder in folder.glob("*"):
        if str(subfolder) == r"D:\ThesisData\front_notrees\1020\GroundTruth" or str(subfolder) == r"D:\ThesisData\front_notrees\322\GroundTruth":
            continue

        data = pd.read_csv(str(subfolder) + r"\data.csv")
        data["grviOut"] = gt_front["grviOut"].iloc[:len(data)]

        dataset = pd.concat([dataset, data], ignore_index=True)

for folder in backPath.glob("*"):
    for subfolder in folder.glob("*"):
        if str(subfolder) == r"D:\ThesisData\back_notrees\1020\GroundTruth" or str(subfolder) == r"D:\ThesisData\back_notrees\322\GroundTruth":
            continue

        data = pd.read_csv(str(subfolder) + r"\data.csv")
        data["grviOut"] = gt_back["grviOut"].iloc[:len(data)]

        dataset = pd.concat([dataset, data], ignore_index=True)
```
