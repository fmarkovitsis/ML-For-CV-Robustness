```python
with open(str(pathToFile), "r") as f:
    grviFile = f.readlines()
```

```python
data = pd.DataFrame(columns=["C0", "C1", "C2", "C3", "C4", "C5", "C6", "C7", "C8", "C9", "C10", "C11", "C12"])

for line in grviFile:
    values = re.findall(r'-?\d+(?:\.\d+)?(?:[eE][+-]?\d+)?', line)

    data.loc[len(data)] = [float(values[0]), float(values[1]), float(values[2]), float(values[3]), float(values[4]), float(values[5]), float(values[6]), float(values[7]), float(values[8]), float(values[9]), float(values[10]), float(values[11]), float(values[12])]

data
```
