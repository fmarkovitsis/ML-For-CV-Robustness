```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv(r"C:\Users\sol77\Desktop\ML-For-CV-Robustness\datasets\oldWronglyCalculated\notrees.csv")

# Calculate GRVI
denominator = df["meanRefG"] + df["meanRefR"]

df["grviIn"] = np.where(
    denominator != 0,
    (df["meanRefG"] - df["meanRefR"]) / denominator,
    np.nan
)

# Save corrected dataset
df.to_csv(r"C:\Users\sol77\Desktop\ML-For-CV-Robustness\datasets\notrees.csv", index=False)

print(f"Processed {len(df)} rows.")
print("Saved to datasetGTCombined_corrected.csv")
```
